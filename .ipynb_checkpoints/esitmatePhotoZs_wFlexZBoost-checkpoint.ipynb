```python
data_file_path = '/Users/leo/Projects/LBG_cosmology/simulated_catalogs/integrated_catalog_23apr26_noised_19Jun26.pq'

training_fraction = 0.8
reference_band = 'LSST_u'

flexZBoost_parameters = dict(
    zmin              = 0.0,
    zmax              = 3.0,
    nzbins            = 301,
    trainfrac         = 0.75,
    bumpmin           = 0.02,
    bumpmax           = 0.35,
    nbump             = 20,
    sharpmin          = 0.7,
    sharpmax          = 2.1,
    nsharp            = 15,
    max_basis         = 35,
    basis_system      = "cosine",
    regression_params = {"max_depth": 8, "objective": "reg:squarederror"},
)

seed = 42
```

```python
date = "07Jul26_test1"
outputs_directory = f"/pscratch/sd/s/sajkov/analysis_pipeline/runs/{date}"
```

```python
PHOTOMETRY_DeepField_noisy    = tables_io.read(f"{outputs_directory}/PHOTOMETRY_DeepField_noisy_{date}.pq")
PHOTOMETRY_WideFastDeep_noisy = tables_io.read(f"{outputs_directory}/PHOTOMETRY_WideFastDeep_noisy_{date}.pq")
PHOTOZS_DeepField_lePhare     = tables_io.read(f"{outputs_directory}/PHOTOZS_DeepField_lePhare_{date}.pq")['photo-z']
REDSHIFTS_DeepField           = np.load(f"{outputs_directory}/TRUEREDSHIFTS_DeepField.npy")
REDSHIFTS_WideFastDeep        = np.load(f"{outputs_directory}/TRUEREDSHIFTS_WideFastDeep.npy")
```

```python
reference_band = 'LSST_i'

flexZBoost_parameters = dict(
    zmin              = 0.0,
    zmax              = 3.0,
    nzbins            = 301,
    trainfrac         = 0.75,
    bumpmin           = 0.02,
    bumpmax           = 0.35,
    nbump             = 20,
    sharpmin          = 0.7,
    sharpmax          = 2.1,
    nsharp            = 15,
    max_basis         = 35,
    basis_system      = "cosine",
    regression_params = {"max_depth": 8, "objective": "reg:squarederror"},
)

seed = 42
```

```python
M5_DEPTHS_DeepField = {'LSST_u'    : 27.74,
                       'LSST_g'    : 28.69,
                       'LSST_r'    : 28.88,
                       'LSST_i'    : 28.96,
                       'LSST_z'    : 28.26,
                       'LSST_y'    : 26.63}

M5_DEPTHS_WideFastDeep = {'LSST_u'    : 25.61,
                          'LSST_g'    : 26.90,
                          'LSST_r'    : 26.87,
                          'LSST_i'    : 26.43,
                          'LSST_z'    : 25.73,
                          'LSST_y'    : 24.79}

bands = M5_DEPTHS_DeepField.keys()
error_bands = [f"{band}_err" for band in bands]
```

```python
TRAINING_DATA_flexZBoost = PHOTOMETRY_DeepField_noisy.copy()
TRAINING_DATA_flexZBoost["redshift"] = REDSHIFTS_DeepField
```

```python
flexZBoost_informer = ri.estimation.algos.flexzboost.flex_z_boost_informer(
    training_data = TRAINING_DATA_flexZBoost, hdf5_groupname = "",
    bands = bands, err_bands = error_bands,
    ref_band = reference_band, redshift_col = 'redshift',
    mag_limits = M5_DEPTHS_DeepField,
    **flexZBoost_parameters
)
```

```python
flexZBoost_estimator = ri.estimation.algos.flexzboost.flex_z_bo ost_estimator(
    input_data = PHOTOMETRY_WideFastDeep_noisy, hdf5_groupname = "",
    model = flexZBoost_informer['model'],
    bands = bands, err_bands = error_bands, ref_band = reference_band,
    mag_limits = M5_DEPTHS_WideFastDeep
)
```

```python
PHOTOZS_flexZBoost = np.reshape(flexZBoost_estimator["output"].median(), len(PHOTOMETRY_WideFastDeep_noisy))
```

```python
fig = plt.figure()
ax = fig.add_subplot()

ax.scatter(REDSHIFTS_WideFastDeep, PHOTOZS_flexZBoost, color = "black", s = 2)

ax.axline((0, 0), slope = 1, color = "red")
ax.axline((0.15, 0), slope = 0.85, color = "red", linestyle = "--", linewidth = 2)
ax.axline((0, 0.15), slope = 1.15, color = "red", linestyle = "--", linewidth = 2)

ax.set_xlabel("True redshift")
ax.set_ylabel("Photometric redshift")
```
